```python
def replace(input_file, output_file, threshold=3):
    with open(input_file, "r", encoding="utf-8") as f_in, open(output_file, "w", encoding="utf-8") as f_out:
        for line in f_in:
            line = line.lstrip()
            semicolon_count = line.count(";")
            if semicolon_count > threshold:
                parts = line.strip().split(";")
                if len(parts) > 2:
                    between = ";".join(parts[2:-1])
                    between_clean = between.replace(";", " ")
                    new_line = ";".join(parts[:2] + [between_clean] + [parts[-1]]) + "\n"
                    f_out.write(new_line)
                else:
                    f_out.write(line )
            else:
                f_out.write(line )
```

```python
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import pandas as pd
import nltk
import re
import numpy as np

nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')

lemmatizer = WordNetLemmatizer()
stop_words = set(stopwords.words('english'))

def preprocess_text(text):
    if not isinstance(text, str):
        return ""
    text = text.lower()
    text = re.sub(r'[^a-z\s]', '', text)
    tokens = nltk.word_tokenize(text)
    tokens = [word for word in tokens if word not in stop_words]
    tokens = [lemmatizer.lemmatize(word) for word in tokens]
    return ' '.join(tokens)

def split_columns(df, column_name):
    if column_name in df.columns:
        df[['index', 'title', 'text', 'label']] = df[column_name].str.split(';', n=3, expand=True)
        df = df.drop(columns=[column_name])
    return df

train_df = split_columns(train_df, ';title;text;label')
test_df = split_columns(test_df, ';title;text;label')
eval_df = split_columns(eval_df, ';title;text;label')

# Convert label column to numeric, coercing errors to NaN
train_df['label'] = pd.to_numeric(train_df['label'], errors='coerce')
test_df['label'] = pd.to_numeric(test_df['label'], errors='coerce')
eval_df['label'] = pd.to_numeric(eval_df['label'], errors='coerce')

# Filter out rows where label is NaN or not a number/integer before astype(int)
train_df = train_df[pd.notna(train_df['label']) & (train_df['label'].apply(lambda x: isinstance(x, (int, float))))]
test_df = test_df[pd.notna(test_df['label']) & (test_df['label'].apply(lambda x: isinstance(x, (int, float))))]
eval_df = eval_df[pd.notna(eval_df['label']) & (eval_df['label'].apply(lambda x: isinstance(x, (int, float))))]

# Convert label to integer
train_df['label'] = train_df['label'].astype(int)
test_df['label'] = test_df['label'].astype(int)
eval_df['label'] = eval_df['label'].astype(int)


train_df['cleaned_text'] = train_df['text'].apply(preprocess_text)
test_df['cleaned_text'] = test_df['text'].apply(preprocess_text)
eval_df['cleaned_text'] = eval_df['text'].apply(preprocess_text)

display(train_df.head())
display(test_df.head())
display(eval_df.head())
```

```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd

tfidf_vectorizer = TfidfVectorizer(max_features=5000)
train_tfidf = tfidf_vectorizer.fit_transform(train_df['cleaned_text'])
train_tfidf_filtered = train_tfidf
train_labels_filtered = train_df['label']

model = MultinomialNB()
model.fit(train_tfidf_filtered, train_labels_filtered)
print("Model training complete.")
```

```python
import shap

# For SHAP, if using a non-tree model, we will use the KernelExplainer.
# We need a background dataset for KernelExplainer.
# It's common to use a sample of the training data for this.

# Create a background dataset for SHAP KernelExplainer
# Taking a small sample to speed up computation. You might adjust this size.
background_data = train_df['cleaned_text'].sample(100, random_state=42).tolist()
background_tfidf = tfidf_vectorizer.transform(background_data)

# Initialize JS (JavaScript) visualization in notebooks
shap.initjs()

# Initialize the KernelExplainer
# The explainer needs a prediction function that takes raw input (texts) and returns probabilities,
# and a background dataset (vectorized form).
explainer = shap.KernelExplainer(predict_proba_for_explainer, background_tfidf)

# Select a sample of test data to explain
x_test_sample = test_df['text'].sample(5, random_state=42)
x_test_sample_cleaned = x_test_sample.apply(preprocess_text)
x_test_sample_vectorized = tfidf_vectorizer.transform(x_test_sample_cleaned)

# Compute SHAP values for the sample
# The predict_proba_for_explainer function expects a list of raw texts.
shap_values = explainer.shap_values(x_test_sample_vectorized)

print("SHAP values calculated.")
```
